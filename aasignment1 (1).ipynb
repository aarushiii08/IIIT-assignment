```python
#7
import pandas as pd

df = pd.read_csv("sales_data.csv")

df["PurchaseDate"] = pd.to_datetime(df["PurchaseDate"], errors="coerce")

df["Year"] = df["PurchaseDate"].dt.year
df["Month"] = df["PurchaseDate"].dt.month
df["Day"] = df["PurchaseDate"].dt.day

print(df[["PurchaseDate", "Year", "Month", "Day"]].head())
```

```python
#11
df = pd.read_csv("sales_data.csv")
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Rating"].fillna(df["Rating"].mean(), inplace=True)
df["DeliveryDays"].fillna(df["DeliveryDays"].median(), inplace=True)
print(df.isnull().sum())
```

```python
#12
import matplotlib.pyplot as plt
df[['Price','Quantity','Age']] = df[['Price','Quantity','Age']].apply(pd.to_numeric, errors='coerce')
df[['Price','Quantity','Age']].boxplot()
plt.show()
for col in ['Price','Quantity','Age']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    print(col)
    print(df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)])
```

```python
#13
before = df.shape[0]
for col in ['Price','Quantity','Age']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    df = df[(df[col] >= Q1 - 1.5*IQR) & (df[col] <= Q3 + 1.5*IQR)]
after = df.shape[0]
print("Dataset size before removal:", before)
print("Dataset size after removal:", after)
```

```python
#14
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Region_Label'] = le.fit_transform(df['Region'])

df = pd.get_dummies(df, columns=['Gender'], dtype=int)

df.head()
```

```python
#15
from sklearn.preprocessing import MinMaxScaler, StandardScaler

mms = MinMaxScaler()
ss = StandardScaler()

df[['Price_MinMax','Quantity_MinMax','Age_MinMax']] = mms.fit_transform(df[['Price','Quantity','Age']])
df[['Price_Std','Quantity_Std','Age_Std']] = ss.fit_transform(df[['Price','Quantity','Age']])

df.head()
```

```python
#16
df['TotalAmount'] = df['Price'] * df['Quantity']
print(df.groupby('ProductCategory')['Price'].mean())
print(df.groupby('Region')['Rating'].mean())
print(df.groupby('PaymentMethod')['TotalAmount'].sum())
```

```python
#17
pd.pivot_table(df, values='TotalAmount', index='Region', columns='ProductCategory', aggfunc='sum')
```

```python
#22
import pandas as pd
import matplotlib.pyplot as plt
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['DeliveryDays'] = pd.to_numeric(df['DeliveryDays'], errors='coerce')
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.boxplot(df['Price'].dropna())
plt.title('Price')

plt.subplot(1,3,2)
plt.boxplot(df['Quantity'].dropna())
plt.title('Quantity')

plt.subplot(1,3,3)
plt.boxplot(df['DeliveryDays'].dropna())
plt.title('DeliveryDays')

plt.tight_layout()
plt.show()
```
